```python
notas = []
soma = 0
for nt in range(5):

  nt = float(input("Digite sua nota:"))
  notas.append(nt)

print(notas)

for nt in notas:
   soma = soma + nt

media = soma / len (notas)

print("A média é:",media)

if media >= 7:
  print(" A turma está com média boa!")

else:
  print("A turma precisa melhorar")
```
